```python
import logging

# Correct and single call to configure logging
logging.basicConfig(
    filename='apple.log',         # log file name
    filemode='w',                 # overwrite each time
    level=logging.DEBUG,          # capture all levels from DEBUG and up
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',  # fixed "message"
    datefmt='%Y-%m-%d %H:%M:%S'
)

# Logging some messages
logging.debug('This is a debug message')
logging.info('This is an info message')
logging.warning('This is a warning message')
logging.error('This is an error message')
logging.critical('This is a critical message')
```

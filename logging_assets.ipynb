```python
import logging 
logger = logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',  # fixed "message"
    datefmt='%Y-%m-%d %H:%M:%S',
    handlers=[
        logging.FileHandler('app2.log'),
        logging.StreamHandler()
    ]
)

logger = logging.getLogger('arhimeticapp')

def mean(a,b,c):
    logger.error('mean is being calculated')
    mean1 = a+b+c
    mean1 = mean1/3
    return mean1

```

```python
mean(1,2,2)
```

```python
logger2 = logging.getLogger('logger2')

def divider(a,b):
    try:
        result = a/b
        logger2.debug('dividing is in progress')
        return result
    except ZeroDivisionError:
        logger2.error('error is happening')


divider(1,0)
        
```

```python
from logging.handlers import RotatingFileHandler
def logger_with_rotating_file_handler():
    logger = logging.getLogger('rotating_logger')
    logger.setLevel(logging.DEBUG)
    

    rotating_filehandler = RotatingFileHandler('rotating_app.log',maxBytes=2000,backupCount=5)
    formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')

    
    rotating_filehandler.setFormatter(formatter)

    logger.addHandler(rotating_filehandler)

    for i in range(100):
        logger.debug(f'this is debug message number {i}')

    
logger_with_rotating_file_handler()
```

```python
list_str = ['salom','qalesn','nimagap']

def list_write(file_path,list_path):
    with open(file_path,'w+') as file:
        for lines in list_path:
            file.write(lines+'\n')
            file.seek(0)
            content = file.read()
    return content 
            
list_write('output.txt',list_str)
```

```python
def copier(cop_file,dest_file):
    with open(cop_file,'r') as copier:
        with open(dest_file ,'w') as dest:
            dest.write(copier.read())
    
    

copier('example.txt','output.txt')

```

```python
def count_words(file):
    with open(file,'r') as file:
        text = file.read()
        words = text.split()
        return len(words)
```

```python
count_words('output.txt')
```

```python
def replcer(filename,new_word,old_word):
    with open(filename,'r') as file:
        content = file.read()
    new_content = content.replace(old_word,new_word)
    with open(filename,'w') as file:
        file.write(new_content)

replcer('sample.txt','saida','aziz')
```

```python

def reverser(filename):
    with open(filename,'r') as file:
        lines = file.readlines()
    with open(filename,'a') as file2:
        for line in reversed(lines):
            file2.write(line)
            
            
        
```

```python
reverser('example.txt')
```

```python
def merger(merger_list,output_file):
    with open(output_file,'w') as file:
        for fname in merger_list:
            with open(fname,'r') as output:
                file.write(output.read()+'\n')
```

```python
merger(['sample.txt','example.txt'],'salom.txt')
```
